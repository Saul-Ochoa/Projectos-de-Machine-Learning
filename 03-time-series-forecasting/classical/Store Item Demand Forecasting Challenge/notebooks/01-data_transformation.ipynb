# **Data Transformation**

Exploración rápida de los CSV crudos del challenge. La limpieza real (parseo de fechas, remoción de duplicados) vive en [`src/data/make_dataset.py`](../src/data/make_dataset.py) y se corre con `python -m src.data.make_dataset` — este notebook solo la reutiliza para inspeccionar el resultado, sin duplicar la lógica.

```python
import sys
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path('..').resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data.make_dataset import clean_dataset, RAW_DIR

ruta_train = (RAW_DIR / "train.csv").resolve()
ruta_test = (RAW_DIR / "test.csv").resolve()
```

```python
# clean_dataset() (src/data/make_dataset.py) parsea 'date' y remueve duplicados;
# aquí solo se inspecciona el resultado, no se duplica la lógica de limpieza.
df_train = clean_dataset(pd.read_csv(ruta_train, sep=','))
df_test = clean_dataset(pd.read_csv(ruta_test, sep=','))

print(f'\n Cantidad de filas y columnas en la data train {df_train.shape}')
print(f'\n Cantidad de filas y columnas en la data test {df_test.shape}')
print(f'\n Total de vacios data train {df_train.isnull().sum()}')
print(f'\n Total de vacios data test {df_test.isnull().sum()}')
```
