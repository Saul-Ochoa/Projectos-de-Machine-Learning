# **Data Transformation**

```python
from pathlib import Path
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Estando en /notebooks, subes un nivel (..)
ruta_features = Path("../data/1.raw/features.csv").resolve()
ruta_stores = Path("../data/1.raw/stores.csv").resolve()
ruta_test = Path("../data/1.raw/test.csv").resolve()
ruta_train = Path("../data/1.raw/train.csv").resolve()
ruta_submission = Path("../data/1.raw/sampleSubmission.csv").resolve()
# print(ruta_features)
# print("Existe?", ruta_features.exists())

df_features = pd.read_csv(ruta_features)
df_stores = pd.read_csv(ruta_stores)
df_test = pd.read_csv(ruta_test)
df_train = pd.read_csv(ruta_train)
df_submission = pd.read_csv(ruta_submission)

df_features=df_features[['Store', 'Date', 'Temperature', 'Fuel_Price', 'MarkDown1', 'MarkDown2',
       'MarkDown3', 'MarkDown4', 'MarkDown5', 'CPI', 'Unemployment']]
df_inner_train=df_train.merge(df_stores,how='left').merge(df_features,how='left')
df_inner_test=df_test.merge(df_stores,how='left').merge(df_features,how='left')
ruta_inner_train = Path("../data/2.processed/processed_train.csv").resolve()
ruta_inner_test = Path("../data/2.processed/processed_test.csv").resolve()
df_inner_test.to_csv(ruta_inner_test,index=False)
df_inner_train.to_csv(ruta_inner_train,index=False)
```
